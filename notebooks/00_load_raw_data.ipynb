```python
import duckdb

conn = duckdb.connect(
    "../customer_support.duckdb"
)

print("Connected successfully")
```

```python
conn.sql("""
CREATE SCHEMA IF NOT EXISTS raw;
""")

print("Raw schema created")
```

```python
conn.sql("""
CREATE OR REPLACE TABLE raw.customer_support_tickets AS

SELECT *
FROM read_csv_auto(
    '../raw_data/customer_support_tickets.csv'
);
""")

print("Raw table created")
```

```python
conn.sql("""
SHOW TABLES FROM raw;
""").df()
```

```python
conn.sql("""
SELECT *
FROM raw.customer_support_tickets
LIMIT 5;
""").df()
```

```python
conn.close()
```
