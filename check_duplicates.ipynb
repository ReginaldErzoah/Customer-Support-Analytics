```python
import duckdb

query = """
select
    customer_email,
    count(*) as records
from main.dim_customers
group by customer_email
having count(*) > 1
order by records desc;
"""

conn = duckdb.connect("customer_support.duckdb")

print(conn.sql(query))
```
